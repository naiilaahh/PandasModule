# Aggregate Functions RLAB

```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Category': ['A', 'A', 'B', 'B', 'A', 'B'],
    'Value': [10, 15, 20, 25, 30, 35],
    'Fee' :[20000,25000,26000,22000,24000,35000],
    'Duration':['30day','40days','35days','40days','60days','60days'],
    'Discount':[1000,2300,1200,2500,2000,2000]
}

df = pd.DataFrame(data)
```

```python
df.head()
```

```python
total_sum = df['Value'].aggregate('sum')
total_sum
```

```python
total_sum2 = df['Value'].sum()
total_sum2
```

```python
# cumulative sum/running total
df[['Value']].cumsum()
```

```python
total_mean = df['Value'].aggregate('mean')
total_mean
```

```python
total_mean2 = df['Value'].mean()

total_mean2
```

```python
total_max= df['Value'].aggregate('max')
total_max
```

```python
total_max2 = df['Value'].max()
total_max2
```

```python
total_min= df['Value'].aggregate('min')
total_min
```

```python
total_min2 = df['Value'].min()
total_min2
```

```python
total_std = df['Value'].aggregate('std')
total_count = df['Value'].aggregate('count')

print(total_count)
print(total_std)
```

```python
total_std2 = df['Value'].std()
total_count2 = df['Value'].count()

print(total_count2)
print(total_std2)
```

#### Applying multiple aggregates

```python
result = df[['Fee', 'Discount']].agg('sum')
result
```

```python
# Determine the total of eachg group with and agg function
result_group = df.groupby('Category')[['Fee', 'Discount']].agg('sum')

print(result_group)
```
