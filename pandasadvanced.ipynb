### Combining series and dataframes
- `pd.concat()` : Allows to add dataframes together of similar shapes
- `pd.append()`
- `pd.merge()`
- `pd.join()`

```python
import pandas as pd
import numpy as nb
```

```python
df1 = pd.DataFrame(
    {
        "A": ["A0", "A1", "A2", "A3"],
        "B": ["B0", "B1", "B2", "B3"],
        "C": ["C0", "C1", "C2", "C3"],
        "D": ["D0", "D1", "D2", "D3"],
    },

)


df2 = pd.DataFrame(
    {
        "A": ["A4", "A5", "A6", "A7"],
        "B": ["B4", "B5", "B6", "B7"],
        "C": ["C4", "C5", "C6", "C7"],
        "D": ["D4", "D5", "D6", "D7"],
    },

)


df3 = pd.DataFrame(
    {
        "A": ["A8", "A9", "A10", "A11"],
        "B": ["B8", "B9", "B10", "B11"],
        "C": ["C8", "C9", "C10", "C11"],
        "D": ["D8", "D9", "D10", "D11"],
    },

)
```

```python
frames = [df1, df2, df3]
# ignore_index to create a new index starting from 0 to end
result = pd.concat(frames, ignore_index=True)
```

```python
result
```

```python
# keys= delineates the different concatenated df
# allows tracking of original df
result = pd.concat(frames, keys=['X', 'Y', 'Z'])
```

```python
result
```

```python
# extracts the original df objects
result_y = result.loc[('Y')]

result_y
```

```python
df4 = pd.DataFrame(
    {
        "B": ["B2", "B3", "B6", "B7"],
        "D": ["D2", "D3", "D6", "D7"],
        "F": ["F2", "F3", "F6", "F7"],
    },
    index=[12, 13, 14, 15],
)

df4
```

```python
result = pd.concat((df1, df4), ignore_index=True, sort=False)

result
```

```python
result = pd.concat((df1, df4), ignore_index=True, axis=1, sort=False)
result
```

## Merging

```python
left_df = pd.DataFrame({'Courses': ["Spark","PySpark","Python","pandas","Java"],
                    'Fee' : [20000,25000,30000,24000,40000],
                    'Duration':['30day','40days','60days','55days','50days']})

right_df = pd.DataFrame({'Courses': ["Java","PySpark","Python","pandas","Hyperion","html"],
                    'Fee': [20000,25000,30000,24000,40000,4000],
                    'Percentage':['10%','20%','25%','20%','10%','50%']})
```

```python
left_df
```

```python
right_df
```

```python
merged = pd.merge(left_df, right_df)
merged
```

```python
merged = pd.merge(left_df, right_df, on='Courses')
merged
```

```python
merged = pd.merge(left_df, right_df, on=['Courses','Fee'])
merged
```

```python
# similar to sql left join
merged_df = pd.merge(
    left_df,
    right_df,
    how='left',
    left_on=['Courses', 'Fee'],
    right_on=['Courses', 'Fee']
)

merged_df
```

## Joining

```python
technologies = {
    'Courses':["Spark","PySpark","Python","pandas"],
    'Fee' :[20000,25000,22000,30000],
    'Duration':['30days','40days','35days','50days'],
              }
index_labels=['r1','r2','r3','r4']

technologies2 = {
    'Courses':["Spark","Java","Python","Go"],
    'Discount':[2000,2300,1200,2000]
              }
index_labels2=['r1','r6','r3','r5']
```

```python
df1 = pd.DataFrame(technologies, index=index_labels)
df2 = pd.DataFrame(technologies2, index=index_labels2)
```

```python
df1
df2
```

```python
df3 = df1.join(df2, lsuffix='_left', rsuffix='_right')
```

```python
df3
```

```python
# right join
df3 = df1.join(df2, lsuffix='_left', rsuffix='_right', how='inner')
df3
```

```python
# joining df and stting new indexes
df3 = df1.set_index('Courses').join(df2.set_index('Courses'), how='inner')
```

```python
df3
```

#### `.diff()` : difference in value from one cell of a series to the next

```python
np.random.seed(seed=13)

df = pd.DataFrame(data=np.random.normal(loc=70, scale=10, size=(7,3)),
           columns=('San Francisco', 'San Diego', 'Los Angeles'),
            index=['Mon', 'Tues', 'Wed', 'Thurs', 'Fri', 'Sat', 'Sun']
                 )

df = df.round()

df
```

```python
df.diff()
```

```python
df.diff(periods=2)
```

```python
df.diff(axis=1)
```

```python
# Create a DataFrame with daily returns for different stocks
np.random.seed(42)
start_date = datetime.date(2022, 1, 1)
end_date = datetime.date(2022, 1, 10)
date_range = pd.date_range(start_date, end_date, freq='D')
# Create a DataFrame with daily returns for different stocks
data = {
    'Date': np.repeat(date_range, 3),
    'Stock': ['AAPL', 'GOOGL', 'MSFT'] * len(date_range),
    'Return': np.random.normal(loc=0.001, scale=0.01, size=len(date_range) * 3)
}

returns_df = pd.DataFrame(data)
```

```python
returns_df
```

```python
pivot = returns_df.pivot(index='Date', columns='Stock', values='Return')
```
